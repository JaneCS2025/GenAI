## Part 2: Peer review 📊
Put each group's scores (1–5) into `scores` and run to see the results.

```python
import matplotlib.pyplot as plt

criteria = ["creativity", "works well", "prompt design", "honesty about AI", "presentation"]
scores = {
    # "Team name": [creativity, works well, prompt design, honesty about AI, presentation],
    "Team A": [4, 5, 3, 5, 4],
    "Team B": [5, 3, 4, 4, 5],
    # 👉 add more teams
}

for team, s in scores.items():
    print(f"{team}: average {sum(s)/len(s):.1f}")

x = range(len(criteria))
w = 0.8 / len(scores)
for i, (team, s) in enumerate(scores.items()):
    plt.bar([j + i * w for j in x], s, width=w, label=team)
plt.xticks([j + w * (len(scores) - 1) / 2 for j in x], criteria, rotation=15)
plt.ylim(0, 5.5); plt.legend(); plt.title("Peer review"); plt.show()
```

## Part 3: 17-lesson review quiz 🧠

```python
quiz = [
    ("AI contains machine learning, which contains ____ learning", "deep"),
    ("Is Face ID discriminative or generative AI?", "discriminative"),
    ("A language model generates text by predicting the next ____", "token"),
    ("Turning the temperature up makes the AI more cautious or more creative?", "creative"),
    ("A neural network learns by stepping downhill on the loss. What is this called?", "gradient descent"),
    ("Which RNN-family model uses forget, input and output gates?", "lstm"),
    ("What is the mechanism a Transformer uses to decide how related words are?", "attention"),
    ("In self-attention, a word's Query is compared with every other word's ____", "key"),
    ("The hidden 'map' inside an autoencoder where similar things sit close together is called ____ space", "latent"),
    ("When a diffusion model paints, what does it start from?", "noise"),
    ("In a GAN, which part catches the fakes?", "discriminator"),
    ("The four parts of a good prompt: role, task, context and ____", "format"),
    ("What do we call it when an AI confidently makes up facts?", "hallucination"),
    ("Our watermark detector counts what colour of words?", "green"),
]
score = 0
for q, a in quiz:
    ans = input(q + "  ")
    if a.lower() in ans.lower():
        score += 1; print("  ✅")
    else:
        print("  ❌ Answer:", a)
print(f"\nTotal: {score}/{len(quiz)}")
```
