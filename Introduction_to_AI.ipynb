## Part 1: What is AI?
**Artificial Intelligence (AI)** = computer systems that do things which normally need human intelligence: understanding language, recognising pictures, making decisions, solving problems.

### 🪆 An analogy: Russian dolls
AI terms fit inside each other like Russian nesting dolls:
- **AI** (biggest doll): any machine that acts "smart" — even one following hand-written rules
- **Machine Learning (ML)**: AI that **learns rules from data** instead of being told them
- **Deep Learning (DL)**: ML that uses big **neural networks** with many layers (Lesson 4)
- **Generative AI** (smallest doll): deep learning that **creates new content** (Lesson 2 onward)

Let's draw it:

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

fig, ax = plt.subplots(figsize=(6, 6))
layers = [(4.0, "#cfe2ff", "Artificial Intelligence"),
          (3.0, "#b6f0c8", "Machine Learning"),
          (2.0, "#ffe3a3", "Deep Learning"),
          (1.0, "#ffb3b3", "Generative\nAI")]
for r, color, label in layers:
    ax.add_patch(Circle((0, r - 4), r, color=color, ec="gray"))   # circles sit on the same bottom point
    y = -3 if r == 1.0 else 2 * r - 4.45                             # label near the top of each circle
    ax.text(0, y, label, ha="center", va="center", fontsize=12, weight="bold")
ax.set_xlim(-4.2, 4.2); ax.set_ylim(-8.2, 0.2); ax.set_aspect("equal"); ax.axis("off")
plt.show()
```

## Part 2: Three levels of AI
| Level | What it means | Exists today? |
|---|---|---|
| **Narrow AI (weak AI)** | Very good at **one** kind of task (playing chess, translating, recognising faces) | ✅ All AI today, including ChatGPT and Claude, is generally considered narrow AI |
| **General AI (AGI)** | As capable as a human at **almost any** intellectual task, and can learn new ones on its own | ❓ Not yet — scientists disagree about when (or whether) it will arrive |
| **Super AI (ASI)** | Smarter than the best humans at almost everything | ❌ Only in science fiction (for now) |

## Part 4: The old way — a rule-based chatbot 📜
Early AI like **ELIZA (1966)** didn't learn anything. Programmers wrote **rules**: "if the user says X, reply Y". Let's build one.

```python
import random

rules = {
    "hello": ["Hi there!", "Hello! How are you?"],
    "sad": ["I'm sorry you feel sad. What happened?", "That sounds hard. Want to talk about it?"],
    "homework": ["Which subject is the homework for?", "Homework again? You can do it!"],
    "game": ["I love games! What's your favourite?"],
    "bye": ["Goodbye! 👋"],
}

def rule_bot(message):
    for keyword, replies in rules.items():
        if keyword in message.lower():
            return random.choice(replies)
    return "Interesting... tell me more."   # when no rule matches

for msg in ["Hello!", "I feel sad today", "I have maths homework", "What is the capital of Japan?"]:
    print("You:", msg, "\nBot:", rule_bot(msg), "\n")
```

## Part 5: Rules vs learning — a spam filter 📧
Imagine writing rules to catch spam emails: "if it contains *free*, it's spam". But spammers just write *fr3e*! Rules break easily.

**Machine learning** flips this around: we give examples, and the computer **works out the rules itself**.

```python
messages = [
    ("WIN a FREE iPhone now!!! Click here", "spam"),
    ("Congratulations you won a prize, claim your cash", "spam"),
    ("Cheap pills, limited offer, buy now", "spam"),
    ("You are selected for a free gift card, click the link", "spam"),
    ("Urgent: your account will be closed, send your password", "spam"),
    ("Are we still meeting for lunch tomorrow?", "ok"),
    ("Here are the notes from today's science class", "ok"),
    ("Mum says dinner is at 6", "ok"),
    ("Can you send me the homework for maths?", "ok"),
    ("Great game on Saturday, see you at training", "ok"),
]
texts, labels = zip(*messages)

# The rule-based way
def rule_filter(text):
    return "spam" if "free" in text.lower() else "ok"

# The machine-learning way
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
vectorizer = CountVectorizer()
ml_filter = MultinomialNB().fit(vectorizer.fit_transform(texts), labels)

tests = ["Claim your fr3e prize now, click here", "See you at lunch", "Win cash now, urgent offer",
         "Is the science homework due tomorrow?"]
for t in tests:
    print(f"rule: {rule_filter(t):5} | ML: {ml_filter.predict(vectorizer.transform([t]))[0]:5} | {t}")
```

```python
# Peek inside: which words did the ML filter learn are "spammy"?
import numpy as np
words = vectorizer.get_feature_names_out()
spam_i = list(ml_filter.classes_).index("spam")
spamminess = ml_filter.feature_log_prob_[spam_i] - ml_filter.feature_log_prob_[1 - spam_i]
top = np.argsort(spamminess)[::-1][:10]
print("Most spammy words:", [words[i] for i in top])
```
