## 🍊 An analogy first
How did you learn to tell apples from oranges when you were little? Nobody gave you a "definition of an apple". You just **saw lots of them**:
red, round, smooth → apple; orange-coloured, bumpy skin → orange.

Machine learning works exactly the same way:
- **Training data** = all the fruit you've seen (with labels: "this is an apple")
- **Features** = what you notice (colour, weight, bumpiness)
- **Model** = the rule your brain worked out
- **Prediction** = seeing a new fruit and saying what it is

```python
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier

# Each fruit: [weight in grams, skin roughness 0–10]
fruits = [[150, 1], [170, 2], [140, 1], [160, 2], [130, 1],   # apples
          [180, 8], [200, 9], [190, 7], [210, 8], [175, 9]]   # oranges
labels = ["apple"] * 5 + ["orange"] * 5

for (w, r), lab in zip(fruits, labels):
    plt.scatter(w, r, c="red" if lab == "apple" else "orange", s=100)
plt.xlabel("weight (g)"); plt.ylabel("roughness"); plt.title("red = apple, orange = orange")
plt.show()
```

```python
# Train! .fit() means "learn"
model = DecisionTreeClassifier()
model.fit(fruits, labels)

# Predict some new fruit
new_fruits = [[155, 2], [195, 8], [170, 5]]
for f, p in zip(new_fruits, model.predict(new_fruits)):
    print(f, "→", p)
```

## Part 2: Recognising handwritten digits ✏️
1,797 tiny 8×8-pixel pictures of handwritten digits. Each picture = 64 numbers (how dark each pixel is).

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

digits = load_digits()
fig, axes = plt.subplots(1, 8, figsize=(10, 2))
for ax, img, lab in zip(axes, digits.images, digits.target):
    ax.imshow(img, cmap="gray_r"); ax.set_title(lab); ax.axis("off")
plt.show()
print("To the computer, the first picture is just these numbers:\n", digits.images[0])
```

**Key idea: training set vs test set**
It's like an exam. The teacher uses practice questions to teach you (training set), but the exam uses **new questions you haven't seen** (test set). Otherwise you could just memorise the answers.

```python
X_train, X_test, y_train, y_test = train_test_split(digits.data, digits.target, test_size=0.3, random_state=0)
knn = KNeighborsClassifier(n_neighbors=3)   # "find the 3 most similar pictures and take a vote"
knn.fit(X_train, y_train)
print(f"Exam score (accuracy): {knn.score(X_test, y_test):.1%}")
```

```python
# Which ones did the AI get wrong?
pred = knn.predict(X_test)
wrong = [i for i in range(len(pred)) if pred[i] != y_test[i]]
fig, axes = plt.subplots(1, min(8, len(wrong)), figsize=(10, 2), squeeze=False)
for ax, i in zip(axes.flat, wrong):
    ax.imshow(X_test[i].reshape(8, 8), cmap="gray_r")
    ax.set_title(f"true:{y_test[i]} guess:{pred[i]}", fontsize=8); ax.axis("off")
plt.show()
```

### 🚀 Challenge
Use the code below to plot a "amount of data vs accuracy" curve.

```python
sizes = [0.98, 0.95, 0.9, 0.7, 0.5, 0.3]
accs = []
for t in sizes:
    Xa, Xb, ya, yb = train_test_split(digits.data, digits.target, test_size=t, random_state=0)
    accs.append(KNeighborsClassifier(3).fit(Xa, ya).score(Xb, yb))
plt.plot([1 - t for t in sizes], accs, "o-")
plt.xlabel("fraction of data used for training"); plt.ylabel("accuracy"); plt.show()
```

## 📌 Summary
- Machine learning = working out the rule from **labelled examples**; nobody writes the rule
- **Too little** data, **wrongly labelled** data, or **one-sided** data → the AI learns the wrong thing
- Generative AI learns the same way — except its data is **a huge chunk of the internet's text and images**
